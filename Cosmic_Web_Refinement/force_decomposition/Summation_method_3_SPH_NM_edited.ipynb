```python
# Boundary integrals
def bound_int(integrand,radius):
    # Here the integrand is the array after being taken the transpose and picked up for the first and the last z bin
    phi_direction = np.sum(integrand * Delta_phi, axis = 0) # So now the phi is the first axis
    r_idx = list(r_bin).index(radius)
    r_direction = np.sum(phi_direction[:r_idx + 1] * Delta_r) # Should be no need to specify the axis now
    return r_direction

# Thermal term, in the axial direction
M_therm_a_1 = np.zeros(rbin_num) # a stands for axial
dP_over_dz_1 = np.gradient(pres_dumped_1, Delta_z, axis = 2)
therm_intergrand_top = np.transpose(rho_inv_1)[-1] * np.transpose(dP_over_dz_1)[-1] * np.transpose(r_dumped_1)[-1]
therm_intergrand_base = np.transpose(rho_inv_1)[0] * np.transpose(dP_over_dz_1)[0] * np.transpose(r_dumped_1)[0]
for i in range(rbin_num):
    radius = r_bin[i]
    M_therm_a_1[i] = bound_int(therm_intergrand_top, radius) - bound_int(therm_intergrand_base, radius)
    M_therm_a_1[i] /=  - 4 * np.pi * gravity

# Flowing term
M_flow_1_1, M_flow_1_2, M_flow_1_3 = np.zeros(rbin_num), np.zeros(rbin_num), np.zeros(rbin_num)
dvz_over_dr_1 = np.gradient(vz_dumped_1, Delta_r, axis = 0)
dvz_over_dphi_1 = np.gradient(vz_dumped_1, Delta_phi, axis = 1)
dvz_over_dz_1 = np.gradient(vz_dumped_1, Delta_z, axis = 2)
flow_integrand_top_1_1 = np.transpose(vr_dumped_1)[-1, :, :] * np.transpose(dvz_over_dr_1)[-1, :, :] * np.transpose(r_dumped_1)[-1, :, :] 
flow_integrand_top_1_2 = np.transpose(vphi_dumped_1)[-1] * np.transpose(dvz_over_dphi_1)[-1]
flow_integrand_top_1_3 = np.transpose(vz_dumped_1)[-1] * np.transpose(dvz_over_dz_1)[-1] * np.transpose(r_dumped_1)[-1]
flow_integrand_base_1_1 = np.transpose(vr_dumped_1)[0] * np.transpose(dvz_over_dr_1)[0] * np.transpose(r_dumped_1)[0] 
flow_integrand_base_1_2 = np.transpose(vphi_dumped_1)[0] * np.transpose(dvz_over_dphi_1)[0]
flow_integrand_base_1_3 = np.transpose(vz_dumped_1)[0] * np.transpose(dvz_over_dz_1)[0] * np.transpose(r_dumped_1)[0]
for i in range(len(r_bin)):
    radius = r_bin[i]
    M_flow_1_1[i] = (bound_int(flow_integrand_top_1_1, radius) - bound_int(flow_integrand_base_1_1, radius)) / (- 4 * np.pi * gravity)
    M_flow_1_2[i] = (bound_int(flow_integrand_top_1_2, radius) - bound_int(flow_integrand_base_1_2, radius)) / (- 4 * np.pi * gravity)
    M_flow_1_3[i] = (bound_int(flow_integrand_top_1_3, radius) - bound_int(flow_integrand_base_1_3, radius)) / (- 4 * np.pi * gravity)
M_flow_1 = M_flow_1_1 + M_flow_1_2 + M_flow_1_3
```

```python
# Boundary integrals
def bound_int(integrand,radius):
    # Here the integrand is the array after being taken the transpose and picked up for the first and the last z bin
    phi_direction = np.sum(integrand * Delta_phi, axis = 0) # So now the phi is the first axis
    r_idx = list(r_bin).index(radius)
    r_direction = np.sum(phi_direction[:r_idx + 1] * Delta_r) # Should be no need to specify the axis now
    return r_direction

# Thermal term, in the axial direction
M_therm_a_3 = np.zeros(len(r_bin)) # a stands for axial
dP_over_dz_3 = np.gradient(pres_dumped_3, Delta_z, axis = 2)
therm_intergrand_top = np.transpose(rho_inv_3)[-1] * np.transpose(dP_over_dz_3)[-1] * np.transpose(r_dumped_3)[-1]
therm_intergrand_base = np.transpose(rho_inv_3)[0] * np.transpose(dP_over_dz_3)[0] * np.transpose(r_dumped_3)[0]
for i in range(len(r_bin)):
    radius = r_bin[i]
    M_therm_a_3[i] = bound_int(therm_intergrand_top, radius) - bound_int(therm_intergrand_base, radius)
    M_therm_a_3[i] /=  - 4 * np.pi * gravity

# Flowing term
M_flow_3_1, M_flow_3_2, M_flow_3_3 = np.zeros(len(r_bin)), np.zeros(len(r_bin)), np.zeros(len(r_bin))
dvz_over_dr_3 = np.gradient(vz_dumped_3, Delta_r, axis = 0)
dvz_over_dphi_3 = np.gradient(vz_dumped_3, Delta_phi, axis = 1)
dvz_over_dz_3 = np.gradient(vz_dumped_3, Delta_z, axis = 2)
flow_integrand_top_3_1 = np.transpose(vr_dumped_3)[-1] * np.transpose(dvz_over_dr_3)[-1] * np.transpose(r_dumped_3)[-1] 
flow_integrand_top_3_2 = np.transpose(vphi_dumped_3)[-1] * np.transpose(dvz_over_dphi_3)[-1]
flow_integrand_top_3_3 = np.transpose(vz_dumped_3)[-1] * np.transpose(dvz_over_dz_3)[-1] * np.transpose(r_dumped_3)[-1]
flow_integrand_base_3_1 = np.transpose(vr_dumped_3)[0] * np.transpose(dvz_over_dr_3)[0] * np.transpose(r_dumped_3)[0] 
flow_integrand_base_3_2 = np.transpose(vphi_dumped_3)[0] * np.transpose(dvz_over_dphi_3)[0]
flow_integrand_base_3_3 = np.transpose(vz_dumped_3)[0] * np.transpose(dvz_over_dz_3)[0] * np.transpose(r_dumped_3)[0]
for i in range(len(r_bin)):
    radius = r_bin[i]
    M_flow_3_1[i] = (bound_int(flow_integrand_top_3_1, radius) - bound_int(flow_integrand_base_3_1, radius)) / (- 4 * np.pi * gravity)
    M_flow_3_2[i] = (bound_int(flow_integrand_top_3_2, radius) - bound_int(flow_integrand_base_3_2, radius)) / (- 4 * np.pi * gravity)
    M_flow_3_3[i] = (bound_int(flow_integrand_top_3_3, radius) - bound_int(flow_integrand_base_3_3, radius)) / (- 4 * np.pi * gravity)
M_flow_3 = M_flow_3_1 + M_flow_3_2 + M_flow_3_3
```

```python
# Acceleration terms
slice_idx = 3

phi_grid = np.zeros((len(r_bin), len(phi_bin), len(z_bin))) # phi value for each bin
for i in range(len(r_bin)):
    for j in range(len(phi_bin)):
        for k in range(len(z_bin)):
            phi_grid[i][j][k] = phi_bin[j]

Delta_CoM_vx_3 = CoM_v_99_list[slice_idx - 1][0] - CoM_v_98_list[slice_idx - 1][0]
Delta_CoM_vy_3 = CoM_v_99_list[slice_idx - 1][1] - CoM_v_98_list[slice_idx - 1][1]
Delta_CoM_vz_3 = CoM_v_99_list[slice_idx - 1][2] - CoM_v_98_list[slice_idx - 1][2]
Delta_CoM_vr_3 = Delta_CoM_vx_3 * np.cos(phi_grid) + Delta_CoM_vy_3 * np.sin(phi_grid)
dvr_over_dt_3 = (vr_dumped_3 - vr_98_dumped_3 + Delta_CoM_vr_3) / (t_99 - t_98)
dvz_over_dt_3 = (vz_dumped_3 - vz_98_dumped_3 + Delta_CoM_vz_3) / (t_99 - t_98)

M_accel_r_3 = 1 / (-4 * np.pi * gravity) * np.sum(np.sum(dvr_over_dt_3 * r_dumped_3 * Delta_z, axis = 2) * Delta_phi, axis = 1)
accel_a_integrand_top = np.transpose(dvz_over_dt_3)[-1] * np.transpose(r_dumped_3)[-1]
accel_a_integrand_base = np.transpose(dvz_over_dt_3)[0] * np.transpose(r_dumped_3)[0]
M_accel_a_3 = np.zeros(len(r_bin))
for i in range(len(r_bin)):
    radius = r_bin[i]
    M_accel_a_3[i] = 1 / (-4 * np.pi * gravity) * (bound_int(accel_a_integrand_top, radius) - bound_int(accel_a_integrand_base, radius))

M_accel_3 = M_accel_r_3 + M_accel_a_3
```

```python
# Boundary integrals
def bound_int(integrand,radius):
    # Here the integrand is the array after being taken the transpose and picked up for the first and the last z bin
    phi_direction = np.sum(integrand * Delta_phi, axis = 0) # So now the phi is the first axis
    r_idx = list(r_bin).index(radius)
    r_direction = np.sum(phi_direction[:r_idx + 1] * Delta_r) # Should be no need to specify the axis now
    return r_direction

# Thermal term, in the axial direction
M_therm_a_5 = np.zeros(len(r_bin)) # a stands for axial
dP_over_dz_5 = np.gradient(pres_dumped_5, Delta_z, axis = 2)
therm_intergrand_top = np.transpose(rho_inv_5)[-1] * np.transpose(dP_over_dz_5)[-1] * np.transpose(r_dumped_5)[-1]
therm_intergrand_base = np.transpose(rho_inv_5)[0] * np.transpose(dP_over_dz_5)[0] * np.transpose(r_dumped_5)[0]
for i in range(len(r_bin)):
    radius = r_bin[i]
    M_therm_a_5[i] = bound_int(therm_intergrand_top, radius) - bound_int(therm_intergrand_base, radius)
    M_therm_a_5[i] /=  - 4 * np.pi * gravity

# Flowing term
M_flow_5_1, M_flow_5_2, M_flow_5_3 = np.zeros(len(r_bin)), np.zeros(len(r_bin)), np.zeros(len(r_bin))
dvz_over_dr_5 = np.gradient(vz_dumped_5, Delta_r, axis = 0)
dvz_over_dphi_5 = np.gradient(vz_dumped_5, Delta_phi, axis = 1)
dvz_over_dz_5 = np.gradient(vz_dumped_5, Delta_z, axis = 2)
flow_integrand_top_5_1 = np.transpose(vr_dumped_5)[-1] * np.transpose(dvz_over_dr_5)[-1] * np.transpose(r_dumped_5)[-1] 
flow_integrand_top_5_2 = np.transpose(vphi_dumped_5)[-1] * np.transpose(dvz_over_dphi_5)[-1]
flow_integrand_top_5_3 = np.transpose(vz_dumped_5)[-1] * np.transpose(dvz_over_dz_5)[-1] * np.transpose(r_dumped_5)[-1]
flow_integrand_base_5_1 = np.transpose(vr_dumped_5)[0] * np.transpose(dvz_over_dr_5)[0] * np.transpose(r_dumped_5)[0] 
flow_integrand_base_5_2 = np.transpose(vphi_dumped_5)[0] * np.transpose(dvz_over_dphi_5)[0]
flow_integrand_base_5_3 = np.transpose(vz_dumped_5)[0] * np.transpose(dvz_over_dz_5)[0] * np.transpose(r_dumped_5)[0]
for i in range(len(r_bin)):
    radius = r_bin[i]
    M_flow_5_1[i] = (bound_int(flow_integrand_top_5_1, radius) - bound_int(flow_integrand_base_5_1, radius)) / (- 4 * np.pi * gravity)
    M_flow_5_2[i] = (bound_int(flow_integrand_top_5_2, radius) - bound_int(flow_integrand_base_5_2, radius)) / (- 4 * np.pi * gravity)
    M_flow_5_3[i] = (bound_int(flow_integrand_top_5_3, radius) - bound_int(flow_integrand_base_5_3, radius)) / (- 4 * np.pi * gravity)
M_flow_5 = M_flow_5_1 + M_flow_5_2 + M_flow_5_3
```

```python
# Boundary integrals
def bound_int(integrand,radius):
    # Here the integrand is the array after being taken the transpose and picked up for the first and the last z bin
    phi_direction = np.sum(integrand * Delta_phi, axis = 0) # So now the phi is the first axis
    r_idx = list(r_bin).index(radius)
    r_direction = np.sum(phi_direction[:r_idx + 1] * Delta_r) # Should be no need to specify the axis now
    return r_direction

# Thermal term, in the axial direction
M_therm_a_7 = np.zeros(len(r_bin)) # a stands for axial
dP_over_dz_7 = np.gradient(pres_dumped_7, Delta_z, axis = 2)
therm_intergrand_top = np.transpose(rho_inv_7)[-1] * np.transpose(dP_over_dz_7)[-1] * np.transpose(r_dumped_7)[-1]
therm_intergrand_base = np.transpose(rho_inv_7)[0] * np.transpose(dP_over_dz_7)[0] * np.transpose(r_dumped_7)[0]
for i in range(len(r_bin)):
    radius = r_bin[i]
    M_therm_a_7[i] = bound_int(therm_intergrand_top, radius) - bound_int(therm_intergrand_base, radius)
    M_therm_a_7[i] /=  - 4 * np.pi * gravity

# Flowing term
M_flow_7_1, M_flow_7_2, M_flow_7_3 = np.zeros(len(r_bin)), np.zeros(len(r_bin)), np.zeros(len(r_bin))
dvz_over_dr_7 = np.gradient(vz_dumped_7, Delta_r, axis = 0)
dvz_over_dphi_7 = np.gradient(vz_dumped_7, Delta_phi, axis = 1)
dvz_over_dz_7 = np.gradient(vz_dumped_7, Delta_z, axis = 2)
flow_integrand_top_7_1 = np.transpose(vr_dumped_7)[-1] * np.transpose(dvz_over_dr_7)[-1] * np.transpose(r_dumped_7)[-1] 
flow_integrand_top_7_2 = np.transpose(vphi_dumped_7)[-1] * np.transpose(dvz_over_dphi_7)[-1]
flow_integrand_top_7_3 = np.transpose(vz_dumped_7)[-1] * np.transpose(dvz_over_dz_7)[-1] * np.transpose(r_dumped_7)[-1]
flow_integrand_base_7_1 = np.transpose(vr_dumped_7)[0] * np.transpose(dvz_over_dr_7)[0] * np.transpose(r_dumped_7)[0] 
flow_integrand_base_7_2 = np.transpose(vphi_dumped_7)[0] * np.transpose(dvz_over_dphi_7)[0]
flow_integrand_base_7_3 = np.transpose(vz_dumped_7)[0] * np.transpose(dvz_over_dz_7)[0] * np.transpose(r_dumped_7)[0]
for i in range(len(r_bin)):
    radius = r_bin[i]
    M_flow_7_1[i] = (bound_int(flow_integrand_top_7_1, radius) - bound_int(flow_integrand_base_7_1, radius)) / (- 4 * np.pi * gravity)
    M_flow_7_2[i] = (bound_int(flow_integrand_top_7_2, radius) - bound_int(flow_integrand_base_7_2, radius)) / (- 4 * np.pi * gravity)
    M_flow_7_3[i] = (bound_int(flow_integrand_top_7_3, radius) - bound_int(flow_integrand_base_7_3, radius)) / (- 4 * np.pi * gravity)
M_flow_7 = M_flow_7_1 + M_flow_7_2 + M_flow_7_3
```

```python
# Boundary integrals
def bound_int(integrand,radius):
    # Here the integrand is the array after being taken the transpose and picked up for the first and the last z bin
    phi_direction = np.sum(integrand * Delta_phi, axis = 0) # So now the phi is the first axis
    r_idx = list(r_bin).index(radius)
    r_direction = np.sum(phi_direction[:r_idx + 1] * Delta_r) # Should be no need to specify the axis now
    return r_direction

# Thermal term, in the axial direction
M_therm_a_9 = np.zeros(len(r_bin)) # a stands for axial
dP_over_dz_9 = np.gradient(pres_dumped_9, Delta_z, axis = 2)
therm_intergrand_top = np.transpose(rho_inv_9)[-1] * np.transpose(dP_over_dz_9)[-1] * np.transpose(r_dumped_9)[-1]
therm_intergrand_base = np.transpose(rho_inv_9)[0] * np.transpose(dP_over_dz_9)[0] * np.transpose(r_dumped_9)[0]
for i in range(len(r_bin)):
    radius = r_bin[i]
    M_therm_a_9[i] = bound_int(therm_intergrand_top, radius) - bound_int(therm_intergrand_base, radius)
    M_therm_a_9[i] /=  - 4 * np.pi * gravity

# Flowing term
M_flow_9_1, M_flow_9_2, M_flow_9_3 = np.zeros(len(r_bin)), np.zeros(len(r_bin)), np.zeros(len(r_bin))
dvz_over_dr_9 = np.gradient(vz_dumped_9, Delta_r, axis = 0)
dvz_over_dphi_9 = np.gradient(vz_dumped_9, Delta_phi, axis = 1)
dvz_over_dz_9 = np.gradient(vz_dumped_9, Delta_z, axis = 2)
flow_integrand_top_9_1 = np.transpose(vr_dumped_9)[-1] * np.transpose(dvz_over_dr_9)[-1] * np.transpose(r_dumped_9)[-1] 
flow_integrand_top_9_2 = np.transpose(vphi_dumped_9)[-1] * np.transpose(dvz_over_dphi_9)[-1]
flow_integrand_top_9_3 = np.transpose(vz_dumped_9)[-1] * np.transpose(dvz_over_dz_9)[-1] * np.transpose(r_dumped_9)[-1]
flow_integrand_base_9_1 = np.transpose(vr_dumped_9)[0] * np.transpose(dvz_over_dr_9)[0] * np.transpose(r_dumped_9)[0] 
flow_integrand_base_9_2 = np.transpose(vphi_dumped_9)[0] * np.transpose(dvz_over_dphi_9)[0]
flow_integrand_base_9_3 = np.transpose(vz_dumped_9)[0] * np.transpose(dvz_over_dz_9)[0] * np.transpose(r_dumped_9)[0]
for i in range(len(r_bin)):
    radius = r_bin[i]
    M_flow_9_1[i] = (bound_int(flow_integrand_top_9_1, radius) - bound_int(flow_integrand_base_9_1, radius)) / (- 4 * np.pi * gravity)
    M_flow_9_2[i] = (bound_int(flow_integrand_top_9_2, radius) - bound_int(flow_integrand_base_9_2, radius)) / (- 4 * np.pi * gravity)
    M_flow_9_3[i] = (bound_int(flow_integrand_top_9_3, radius) - bound_int(flow_integrand_base_9_3, radius)) / (- 4 * np.pi * gravity)
M_flow_9 = M_flow_9_1 + M_flow_9_2 + M_flow_9_3
```

```python
# Acceleration terms
slice_idx = 9

phi_grid = np.zeros((len(r_bin), len(phi_bin), len(z_bin))) # phi value for each bin
for i in range(len(r_bin)):
    for j in range(len(phi_bin)):
        for k in range(len(z_bin)):
            phi_grid[i][j][k] = phi_bin[j]

Delta_CoM_vx_9 = CoM_v_99_list[slice_idx - 1][0] - CoM_v_98_list[slice_idx - 1][0]
Delta_CoM_vy_9 = CoM_v_99_list[slice_idx - 1][1] - CoM_v_98_list[slice_idx - 1][1]
Delta_CoM_vz_9 = CoM_v_99_list[slice_idx - 1][2] - CoM_v_98_list[slice_idx - 1][2]
Delta_CoM_vr_9 = Delta_CoM_vx_9 * np.cos(phi_grid) + Delta_CoM_vy_9 * np.sin(phi_grid)
dvr_over_dt_9 = (vr_dumped_9 - vr_98_dumped_9 + Delta_CoM_vr_9) / (t_99 - t_98)
dvz_over_dt_9 = (vz_dumped_9 - vz_98_dumped_9 + Delta_CoM_vz_9) / (t_99 - t_98)

M_accel_r_9 = 1 / (-4 * np.pi * gravity) * np.sum(np.sum(dvr_over_dt_9 * r_dumped_9 * Delta_z, axis = 2) * Delta_phi, axis = 1)
M_accel_a_9 = np.zeros(len(r_bin))
accel_a_integrand_top = np.transpose(dvz_over_dt_9)[-1] * np.transpose(r_dumped_9)[-1]
accel_a_integrand_base = np.transpose(dvz_over_dt_9)[0] * np.transpose(r_dumped_9)[0]
for i in range(len(r_bin)):
    radius = r_bin[i]
    M_accel_a_9[i] = (bound_int(Hubble_integrand_top, radius) - bound_int(Hublle_integrand_base, radius)) / (- 4 * np.pi * gravity)
M_accel_9 = M_accel_r_9 + M_accel_a_9
```
